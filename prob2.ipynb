```python
# Problem 2: Water Jug Problem using BFS

from collections import deque

# Capacities of the two jugs
CAPACITY_A = 4
CAPACITY_B = 3


# --------------------------------------------------
# Generate all possible states from a given state
# --------------------------------------------------

def get_next_states(state):
    a, b = state
    next_states = []

    # 1. Fill Jug A completely
    next_states.append(((CAPACITY_A, b), "Fill Jug A"))

    # 2. Fill Jug B completely
    next_states.append(((a, CAPACITY_B), "Fill Jug B"))

    # 3. Empty Jug A
    next_states.append(((0, b), "Empty Jug A"))

    # 4. Empty Jug B
    next_states.append(((a, 0), "Empty Jug B"))

    # 5. Pour Jug A -> Jug B
    amount = min(a, CAPACITY_B - b)
    next_states.append((
        (a - amount, b + amount),
        "Pour Jug A -> Jug B"
    ))

    # 6. Pour Jug B -> Jug A
    amount = min(b, CAPACITY_A - a)
    next_states.append((
        (a + amount, b - amount),
        "Pour Jug B -> Jug A"
    ))

    return next_states


# --------------------------------------------------
# BFS to find the minimum number of steps
# --------------------------------------------------

def water_jug_bfs():
    start = (0, 0)

    queue = deque([start])

    visited = {start}

    # Store previous state and operation
    parent = {
        start: None
    }

    operation = {
        start: None
    }

    while queue:

        current = queue.popleft()
        a, b = current

        # Goal: exactly 2 litres in either jug
        if a == 2 or b == 2:

            path = []

            while current is not None:
                path.append(
                    (current, operation[current])
                )
                current = parent[current]

            path.reverse()
            return path

        # Generate next states
        for next_state, action in get_next_states(current):

            if next_state not in visited:

                visited.add(next_state)
                parent[next_state] = current
                operation[next_state] = action

                queue.append(next_state)

    return None


# --------------------------------------------------
# Display the solution
# --------------------------------------------------

solution = water_jug_bfs()

print("Minimum sequence of steps:\n")

for i, (state, action) in enumerate(solution):

    if i == 0:
        print(f"Step {i}: {state}  (Initial State)")
    else:
        print(f"Step {i}: {state}  <- {action}")

print("\nMinimum number of operations:", len(solution) - 1)
```

```python
# Problem 3: Checking Routes Between Cities using DFS

# --------------------------------------------------
# Road network represented using an adjacency list
# --------------------------------------------------

graph = {
    'Delhi': ['Mumbai', 'Jaipur'],
    'Mumbai': ['Delhi', 'Pune'],
    'Jaipur': ['Delhi'],
    'Pune': ['Mumbai'],

    'Kolkata': ['Bhubaneswar'],
    'Bhubaneswar': ['Kolkata'],

    'Chennai': ['Bangalore'],
    'Bangalore': ['Chennai']
}


# --------------------------------------------------
# 1. Check whether a route exists between two cities
# --------------------------------------------------

def route_exists(graph, start, destination):

    visited = set()

    def dfs(city):

        # Destination found
        if city == destination:
            return True

        visited.add(city)

        for neighbour in graph.get(city, []):

            if neighbour not in visited:

                if dfs(neighbour):
                    return True

        return False

    return dfs(start)


# --------------------------------------------------
# 2. Find all connected components
# --------------------------------------------------

def find_connected_components(graph):

    visited = set()
    components = []

    # Include every city appearing in the graph
    cities = set(graph.keys())

    for neighbours in graph.values():
        cities.update(neighbours)

    for city in cities:

        if city not in visited:

            component = []

            def dfs(node):

                visited.add(node)
                component.append(node)

                for neighbour in graph.get(node, []):

                    if neighbour not in visited:
                        dfs(neighbour)

            dfs(city)

            components.append(component)

    return components


# --------------------------------------------------
# Test route checking
# --------------------------------------------------

start_city = 'Delhi'
destination_city = 'Pune'

if route_exists(graph, start_city, destination_city):
    print(
        f"Route exists between {start_city} "
        f"and {destination_city}."
    )
else:
    print(
        f"No route exists between {start_city} "
        f"and {destination_city}."
    )


# --------------------------------------------------
# Find and display connected components
# --------------------------------------------------

components = find_connected_components(graph)

print("\nConnected Components:")

for i, component in enumerate(components, start=1):

    print(
        f"Component {i}: "
        + " -> ".join(sorted(component))
    )
```
